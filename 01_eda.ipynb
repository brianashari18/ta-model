```
# EDA Base Configuration
import os
import re
import json
import pickle
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams.update({'font.size': 18})
plt.rcParams['axes.titlesize'] = 20
plt.rcParams['axes.labelsize'] = 18
plt.rcParams['xtick.labelsize'] = 16
plt.rcParams['ytick.labelsize'] = 16
plt.rcParams['legend.fontsize'] = 16
import torch

# --- GLOBAL PATHS ---
BASE_PATH = Path('./input')
PROCESSED_PATH = Path('./processed')
OUTPUT_PATH = Path('./outputs')

PROCESSED_PATH.mkdir(exist_ok=True)
OUTPUT_PATH.mkdir(exist_ok=True)

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
torch.manual_seed(RANDOM_SEED)

print(f'Stage: {stage_name}')
print(f'Input Path: {BASE_PATH}')
```

```
def load_data(instrument, feature_type):
    path_pkl = BASE_PATH / instrument / feature_type / f'extracted_features_{feature_type}.pkl'
    path_csv = BASE_PATH / instrument / feature_type / f'extracted_features_{feature_type}.csv'
    
    if path_pkl.exists():
        with open(path_pkl, 'rb') as f:
            return pickle.load(f)
    return pd.read_csv(path_csv)

# Pick a sample for EDA (e.g., raw_audio/mfcc)
df = load_data('raw_audio', 'mfcc')
print(f'Loaded {len(df)} rows sample')
df.head()
```

```
# Label Distribution
plt.figure(figsize=(15, 5))
sns.countplot(data=df, x='roman_numeral', order=df['roman_numeral'].value_counts().index)
plt.title('Distribution of Roman Numeral Chords')
plt.xticks(rotation=45)
plt.show()
```
